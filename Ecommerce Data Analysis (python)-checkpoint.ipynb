## Ecommerce Data Analysis: Introduction

**What I did:**
I took a raw Ecommerce sales dataset, cleaned it, and engineered new features to prepare it for analysis and reporting.

**Why:**
The raw data as collected was inconsistent as it had missing values, duplicates, inconsistent formatting, and incorrect data types and didn't directly answer business questions. Before any meaningful analysis could happen, I needed to clean the data and reshape it into metrics that are actually useful, like revenue, profit, and time-based patterns.

**Questions I explored:**
- How much revenue (sales) was generated per order?
- What was the actual net amount collected after discounts?
- How much profit was made after costs?

##### Removing duplicate rows from the DataFrame and updating it directly

```python
df.drop_duplicates(inplace = True)
```

##### Re-checking the number of duplicate rows to confirm they've been removed

```python
df.duplicated().sum()
```

##### Replacing inconsistent missing values with a standard NaN (This helps standardize missing values in a DataFrame because pandas recognizes NaN as a missing value.)

```python
df.replace(['N/A', 'NULL', ''], np.nan, inplace = True)
```

##### Stripping extra whitespace from text (string) columns 
##### We will remove unwanted leading and trailing spaces from text values so that data is clean and consistent.

```python
df = df.apply(lambda x:x.str.strip() if x.dtype == "str" else x)
```

##### Standardizing customer_name, city, and state to title case

```python
df["customer_name"] = df["customer_name"].str.title()

df["city"] = df["city"].str.title()

df["state"] = df["state"].str.title()
```

##### Filtering out rows with invalid emails, keeping only those containing "@" .

```python
df = df[df["email"].str.contains("@",na = False)]
```

##### Converting 'order_date' and 'delivery_date' to a proper datetime format and convert invalid values to NaT (missing date/time)
##### We are doing this because dates stored as plain text cannot be reliably sorted, filtered, compared, or used for date calculations; converting them to datetime lets pandas perform these operations correctly
###### errors="coerce" prevents invalid dates from causing errors by converting them to NaT.

```python
df["order_date"] = pd.to_datetime(df["order_date"],errors = "coerce")

df["delivery_date"] = pd.to_datetime(df["delivery_date"],errors = "coerce")
```

##### Converting qty, unit_price, and discount into numeric values so we can perform calculations and analysis on them, while converting invalid values to NaN instead of causing errors.

```python
cols = ["qty","unit_price","discount"]

for c in cols:
    df[c] = pd.to_numeric(df[c],errors = "coerce")
```

##### Filtering out rows where quantity is zero or negative, this keeps values more than 0

```python
df = df[df["qty"]>0]
```

##### Filling missing discount values with 0
##### Filling missing phone values with "Unknown"
##### Filling missing delivery_date values with the corresponding order_date (this makes delivery date same as order date)

```python
df["discount"] = df["discount"].fillna(0)

df["phone"] = df["phone"].fillna("Unknown")

df["delivery_date"] = df["delivery_date"].fillna(df["order_date"])
```

#### Starting Feature Engineering here since the data is now clean, and we need to derive business-relevant metrics (sales, profit, time periods) that the raw columns don't directly provide. We will be answering our questions here:

##### Calculating total sales as qty multiplied by unit_price . 
- Question 1: How much revenue (sales) was generated per order?

```python
df['sales'] = df['qty'] * df['unit_price']
```

##### Displaying the full DataFrame

```python
df
```

##### Calculating net_amount as sales minus the discount applied.

- Question 2: What was the actual net amount collected after discounts?

```python
df["net_amount"] = df["sales"]-(df["sales"]*df["discount"]/100)
```

##### Displaying the full DataFrame

```python
df
```

##### Calculating profit as 15% of net_amount (we are assuming profit as 15% here)
- Question 3: How much profit was made after costs?

```python
df["profit"] = df["net_amount"] * 0.15
```

##### Displaying the full DataFrame

```python
df
```


##### Extracting month, year, and weekday from order_date into a new column so we can analyze time-based patterns

```python
df["month"] = df["order_date"].dt.month_name()
df["year"] = df["order_date"].dt.year
df["weekday"] = df["order_date"].dt.day_name()
```

##### Displaying the full DataFrame

```python
df
```
